```python
# Difference: Return a new set with elements
#             in the set that are not in the others.
a - b   # letters in a but not in b
```

```python
# Union: Return a new set with elements
#        from the set and all others.
a | b   # letters in a or b or both
```

```python
# Intersection: Return a new set with elements
#               common to the set and all others. 
a & b   # letters in both a and b
```

```python
# Symmetric_difference: Return a new set with elements
#                       in either the set or other but not both.
a ^ b   # letters in a or b but not both
```

```python
# Statistics.py

def getNum():
    # Get user input
    nums = []
    iNumStr = input("Please input a number (Exit with enter): ")
    while iNumStr != "":
        nums.append(eval(iNumStr))
        iNumStr = input("Please input a number (Exit with enter): ")
    return nums

def mean(numbers):
    # calculate the mean value
    s = 0.0
    for num in numbers:
        s += num
    return s / len(numbers)

def dev(numbers, mean): 
    # calculate the standard deviation
    sdev = 0.0
    for num in numbers:
        sdev += (num - mean) ** 2
    return pow(sdev / (len(numbers)-1), 0.5)

def median(numbers):    
    # Calculate the median
    
    # Sort the elements of the list
    # sorted(iterable, key=None, reverse=False)
    sorted(numbers) 
    
    # get the total number of numbers
    size = len(numbers)
    
    # a // b, floor division 
    if size % 2 == 0:  # even
        med = (numbers[size // 2 - 1] + numbers[size // 2]) / 2
    else:  # odd
        med = numbers[size // 2]
    return med

nums = getNum() 
m = mean(nums)
print("Mean:{}, Dev.:{:.2}, Median:{}.".format(m, dev(nums, m), median(nums)))
```

```python
# Calculator implemented with dict

import math

oper_dict = {
    '+': lambda x, y: x + y,
    '-': lambda x, y: x - y,
    '*': lambda x, y: x * y,
    '/': lambda x, y: x / y,
    'exp': lambda x: math.exp(x),
    'ln': lambda x: math.log(x)
}

while(True):
    print('Input a prefix expression.')
    print('Separate the operator and the numbers by whitespace.')
    print('(Exit with Enter)')
    s = input('Input: ')
    
    if len(s) == 0:
        break
    tokens = s.split(' ')
    operation = oper_dict[tokens[0]]
    operands = []
    for token in tokens[1:]:
        operands.append(float(token))
    print('Result: {0:.4f}'.format(operation(*operands)))
```

```python
# word frequency statistics
# implementation with lists

import jieba

txt = open("bit_brief.txt", "r", encoding='utf-8').read()
words = jieba.lcut(txt)
uniwords = []  # a list for unique words
counts = []   # a list for counts

for w in words:
    if len(w) == 1:
        # if this is a single character, 
        # do nothing
        continue
    elif w in uniwords:
        tmp_ind = uniwords.index(w)
        counts[tmp_ind] += 1
    else:
        uniwords.append(w)
        counts.append(1)

for i in range(len(counts)):
    print('{0:<10}{1:>5}'.format(uniwords[i], counts[i]))
```

```python
# word frequency statistics
# implementation with dictionaries

import jieba

txt = open("bit_brief.txt", "r", encoding='utf-8').read()
words = jieba.lcut(txt)
counts = {}  # a dict for counts

for w in words:
    if len(w) == 1:
        # if this is a single character, 
        # do nothing
        continue
    else:
        # get(key[,d]) 
        # Returns the value of the key. 
        # If the key does not exist, returns d (defaults to None).
        counts[w] = counts.get(w, 0) + 1

items = list(counts.items())
print('The first 2 elements of items->', items[:2])
items.sort(key=lambda x:x[1], reverse=True)
print('After sorted->', items[:2])

topN = 15  # show topN words with the most counts
for i in range(topN):
    word, count = items[i]
    print('{0:<10}{1:>5}'.format(word, count))
```
