```python
# Comparison operators are used to compare values. 
# It returns either True or False according to the condition.
x = 10
y = 12

# Output: x > y is False
print('x > y is', x > y)

# Output: x < y is True
print('x < y is', x < y)

# Output: x == y is False
print('x == y is', x == y)

# Output: x != y is True
print('x != y is', x != y)

# Output: x >= y is False
print('x >= y is', x >= y)

# Output: x <= y is True
print('x <= y is', x <= y)
```

```python
# Logical Operators
x = True
y = False

print('x and y is', x and y)

print('x or y is', x or y)

print('not x is', not x)
```

```python
# Membership operators
x = 'Hello world'
y = {1:'a',2:'b'}

# Output: True
print('H' in x)

# Output: True
print('hello' not in x)

# Output: True
print(1 in y)

# Output: False
print('a' in y)
```

```python
# BMI
height, weight = eval(input("Please input height(m) and weight(kg)[separated by comma]: "))
bmi = weight / pow(height, 2)
print("BMI value is：{:.2f}".format(bmi))
                            
cdc, dom = "", ""
                            
if bmi < 18.5:
    cdc, dom = "Thin", "Thin"
elif 18.5 <= bmi < 24:
    cdc, dom = "Healthy", "Healthy"
elif 24 <= bmi < 25:
    cdc, dom = "Healthy", "Overweight"
elif 25 <= bmi < 28:
    cdc, dom = "Overweight", "Overweight"
elif 28 <= bmi < 30:
    cdc, dom = "Overweight ", "Obese"
else:
    cdc, dom = "Obese", "Obese"
print("Weight Status: international - {}, domestic - {}.".format(cdc, dom))
```

```python
# CalPi_v2.py
from random import random
from time import perf_counter

DARTS = 1000 * 1000* 10
hits = 0.0
start = perf_counter()

for i in range(DARTS):
    x, y = random(), random()
    dist = pow(x ** 2 + y ** 2, 0.5)
    if dist <= 1.0:
        hits = hits + 1
pi = 4 * (hits / DARTS)

print("The value of PI is: {}".format(pi))
print("The runtime is: {:.5f}s".format(perf_counter() - start))
```
