```python
# AutoTraceDraw.py

import turtle as t

# Reading Data
datals = []
f = open("para_draw.txt")
for line in f:
    line = line.replace("\n","")
    datals.append(list(map(eval, line.split(","))))
f.close()

# print(datals)

t.title('Trajectory Drawing')
t.setup(500, 480, 0, 0)
t.penup()
t.fd(-150)
t.pendown()
t.pencolor('red')
t.pensize(5)

# Drawing
for i in range(len(datals)):
    t.pencolor(datals[i][3], datals[i][4], datals[i][5])
    t.fd(datals[i][0])
    if datals[i][1]:
        t.right(datals[i][2])
    else:
        t.left(datals[i][2])
t.done()
```
