```python
# default arguments
print("Hello {}, your balance is {}.".format("Adam", 230.2346))

# positional arguments
print("Hello {0}, your balance is {1}.".format("Adam", 230.2346))

# keyword arguments
print("Hello {name}, your balance is {blc}.".format(name="Adam", blc=230.2346))

# mixed arguments
print("Hello {0}, your balance is {blc}.".format("Adam", blc=230.2346))
```

```python
# Note: In case of mixed arguments, keyword arguments has to always follow positional arguments.

print("Hello {blc}, your balance is {1}.".format(blc=230.2346), "Adam")
```

```python
# integer arguments
print("The number is: {:d}".format(123))

# float arguments
print("The float number is: {:f}".format(123.4567898))

# octal, binary and hexadecimal format
print("bin: {0:b}, oct: {0:o}, hex: {0:x}".format(12))
```

```python
# integer numbers with minimum width

# {:5d} takes an integer argument, and assigns a minimum width of 5. 
# Since, no alignment is specified, it is aligned to the right.

print("{:5d}".format(12))
```

```python
# width doesn't work for numbers longer than padding

# the width (2) is less than the number (1234), 
# so it doesn't take any space to the left but also doesn't truncate the number.

print("{:2d}".format(1234))
```

```python
# padding for float numbers

# {:8.3f} truncates the decimal part into 3 places rounding off the last 2 digits. 
# And, the number, now 12.235, takes a width of 8 as a whole 
# leaving 2 places to the left.

print("{:8.3f}".format(12.2346))
```

```python
# If you want to fill the remaining places with zero, 
# placing a zero before the format specifier does this. 
# It works both for integers and floats:

# integer numbers with minimum width filled with zeros
print("{:05d}".format(12))

print("{:15d}".format(12))

# padding for float numbers filled with zeros
print("{:18.3f}".format(12.2346))
```

```python
# integer numbers with right alignment
print("{:5d}".format(12))

# float numbers with center alignment
print("{:^10.3f}".format(12.2346))

# integer left alignment filled with zeros
print("{:<05d}".format(12))
```

```python
# string padding with left alignment
print("{:5}".format("cat"))

# string padding with right alignment
print("{:>5}".format("cat"))

# string padding with center alignment
print("{:^5}".format("cat"))

# string padding with center alignment
# and '*' padding character
print("{:*^5}".format("cat"))
```

```python
# truncating strings to 3 letters
print("{:.3}".format("caterpillar"))

# truncating strings to 3 letters
# and padding
print("{:5.3}python".format("caterpillar"))

# truncating strings to 3 letters,
# padding and center alignment
print("{:*^5.3}".format("caterpillar"))
```

```python
# Text progree bar
import time
scale = 10
print('--------Start-------')
for i in range(scale + 1):
    star = '**' * i
    dot = '..' * (scale - i)
    prog = (i / scale) * 100
    print('%{0:^3.0f}[{1} -> {2}]'.format(prog, star, dot))
    time.sleep(1)  # try what happens if comment this line

print('--------End-------')
```

```python
# Text progree bar
import time
scale = 50
print('Start'.center(scale//2, '-'))

strt = time.perf_counter()
for i in range(scale + 1):
    star = '*' * i
    dot = '.' * (scale - i)
    prog = (i / scale) * 100
    dur = time.perf_counter() - strt
    print('\r{:^3.0f}%[{} -> {}]{:.2f}s '.format(prog, star, dot, dur), end='')
    time.sleep(0.2)

print('\n' + 'End'.center(scale//2, '-'))
```
