```python
class Student(object):
    def __init__(self, name, score):
        # class constructor
        self.name = name
        self.score = score
        
    def print_score(self):
        # methods
        print('%s: %d' % (self.name, self.score))
        
    def get_grade(self):
        if self.score >= 90:
            return 'A'
        elif self.score >= 60:
            return 'B'
        else:
            return 'F'
```

```python
class Student(object):
    def __init__(self, name, score):
        # class constructor
        self.name = name
        self.score = score

# create a Student object
Tom = Student('Tom', 73)
# create a new attribute 'major'
Tom.major = 'Big Data'
print(Tom.name, Tom.score, Tom.major)

# create a Student object
Jerry = Student('Jerry', 80)
# Jerry object doesn't have attribute 'major'
# AttributeError
print(Jerry.name, Jerry.score, Jerry.major)
```

```python
Jerry = Student('Jerry', 80)
del Jerry.score
# AttributeError
print(Jerry.name, Jerry.score, Jerry.major)
```

```python
class Animal(object):
    def run(self):
        print('Animal is running...')
        
class Dog(Animal):
    pass

class Cat(Animal):
    pass

dog = Dog()
dog.run()

cat = Cat()
cat.run()
```
